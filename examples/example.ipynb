```python
from pytrajlib import plot, simulation
```

```python
run_params = simulation.get_run_params()
run_params["num_runs"] = 10
# run_params["atm_model"] = 1
# run_params["atm_error"] = 0
run_params["impact_output"] = 0
run_params["traj_output"] = 2
impact_data = simulation.run(run_params, aim_lat_lon=[70, 30], launch_lat_lon=[-10, -30])
plot.map(run_params=run_params, data=impact_data)
```

```python
plot.impact(run_params=run_params, data=impact_data)
```
