```python
from pytrajlib import plot, simulation
```

```python
run_params = simulation.get_run_params()
run_params["num_runs"] = 100
run_params["impact_output"] = 0
run_params["traj_output"] = 2
impact_data = simulation.run(run_params, launch_lat_lon=[-10, -30], aim_lat_lon=[70, 10])
plot.map(run_params=run_params, data=impact_data, show_attribution=True)
```

```python
plot.impact(run_params=run_params, data=impact_data)
```

```python
plot.all_trajectory_plots(run_params)
```
